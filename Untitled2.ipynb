```python
# Import PyTorch and necessary modules for image processing and pipelines
import torch
from diffusers import StableDiffusionImg2ImgPipeline
from PIL import Image
import requests
from io import BytesIO

# Set device to GPU (cuda) for fast generation
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# Define a stable pre-trained model capable of image-to-image translation
model_id = "runwayml/stable-diffusion-v1-5"

# Load the image-to-image pipeline with float16 optimization for Colab's GPU
pipe = StableDiffusionImg2ImgPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

# Move the pipeline to the GPU
pipe = pipe.to(device)
print("Image-to-Image pipeline loaded successfully!")
```

```python
# Download a sample image to use as input for image-to-image translation
url = "https://raw.githubusercontent.com/CompVis/stable-diffusion/main/assets/stable-samples/img2img/sketch-mountains-input.jpg"
response = requests.get(url)
init_image = Image.open(BytesIO(response.content)).convert("RGB")

# Resize the image to 512x512 pixels for the model
init_image = init_image.resize((512, 512))

# Display the input image in Colab
print("Input Image (e.g., a simple sketch):")
display(init_image)
```

```python
# Write a prompt describing how you want to transform the input image
prompt = "A fantasy landscape with vibrant colors, highly detailed, professional digital painting"

# Perform image-to-image translation using the model
# strength controls how much the model changes the original image (0.0 to 1.0)
images = pipe(prompt=prompt, image=init_image, strength=0.75, guidance_scale=7.5).images

# Get the first generated output image
result_image = images[0]

# Display the translated output image
print("Translated Output Image:")
display(result_image)
```
